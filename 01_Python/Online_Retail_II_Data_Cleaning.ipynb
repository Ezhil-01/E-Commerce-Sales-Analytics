### Note

The **Customer ID** column contains missing values because the Online Retail II dataset includes guest customer transactions.

These records were intentionally retained, as they contribute to revenue and order analysis. Customer-based metrics were calculated using only valid Customer IDs.

```python
df = df[df["Description"].notna()]
```

```python
df["Description"].isnull().sum()
```

```python
df = df[df["Country"].notna()]
```

```python
df["Country"].isnull().sum()
```

```python
df = df[df["Price"].notna()]
```

```python
df["Price"].isnull().sum()
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
```

```python
df = df[df["Quantity"]>0]
```
